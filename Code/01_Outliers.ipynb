**Import NumPy**

```python
import numpy as shafin
```

```python
arr1=shafin.array([2,3,4,6,7,8,9,12,13,16,17,23,25,27,34,37,201])
arr1
```

**Quartile**

```python
Q1=shafin.percentile(arr1,25)
Q2=shafin.percentile(arr1,50)
Q3=shafin.percentile(arr1,75)
print(Q1)
print(Q2)
print(Q3)
```

```python
IQR= Q3-Q1
print(IQR)
```

```python
Upper_Fence = Q3 + (1.5*IQR)
Lower_Fence = Q1 - (1.5*IQR)
print(Upper_Fence)
print(Lower_Fence)
```

```python
l = []    # List
for i in arr1:
    if i >= Lower_Fence and i <= Upper_Fence:
        l.append(i)
arr2 = shafin.array(l)
print (arr2)
```
